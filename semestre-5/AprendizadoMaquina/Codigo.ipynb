```python
# ============================================================
# 1. IMPORTS
# ============================================================

import os
import numpy as np
import pandas as pd

import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.preprocessing import StandardScaler
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
from sklearn.linear_model import LinearRegression

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Dense
from tensorflow.keras.callbacks import EarlyStopping


# ============================================================
# 2. CARREGANDO O DATASET
# ============================================================
bitcoin_df = pd.read_csv('/content/sample_data/coin_Bitcoin.csv')
print("Dataset carregado com sucesso!")
display(bitcoin_df.head())
display(bitcoin_df.describe())
print("Colunas:", bitcoin_df.columns)


# ============================================================
# 3. ANALISE INICIAL (DISTRIBUIÇÃO E OUTLIERS)
# ============================================================
sns.histplot(data=bitcoin_df, x='Close', kde=True)
plt.title('Distribution of Close Price')
plt.show()

sns.boxplot(data=bitcoin_df, x='Close')
plt.title('Box Plot de Close Price')
plt.show()

print("Valores nulos:\n", bitcoin_df.isna().sum())
print("Duplicados:", bitcoin_df.duplicated().sum())


# ============================================================
# 4. MATRIZ DE CORRELAÇÃO
# ============================================================
numeric_df = bitcoin_df.drop(columns=['SNo', 'Name', 'Symbol', 'Date'])
correlation_matrix = numeric_df.corr()

plt.figure(figsize=(10, 8))
sns.heatmap(correlation_matrix, annot=True, fmt='.2f', cmap='coolwarm', linewidths=0.5)
plt.title('Correlation Heatmap')
plt.savefig('bitcoin_correlation_heatmap.png')
plt.show()

print("Heatmap salvo.")


# ============================================================
# 5. TRATAMENTO DE DATA / LIMPEZA
# ============================================================
bitcoin_df = bitcoin_df.drop(columns=['SNo', 'Name', 'Symbol'])
bitcoin_df['Date'] = pd.to_datetime(bitcoin_df['Date'])
bitcoin_df = bitcoin_df.sort_values(by='Date')
display(bitcoin_df.info())


# ============================================================
# 6. SKEWNESS E HISTOGRAMAS DAS FEATURES
# ============================================================
columns_to_check = ['High', 'Low', 'Open', 'Close', 'Volume', 'Marketcap']

print("--- Skewness ---")
print(bitcoin_df[columns_to_check].skew())

bitcoin_df[columns_to_check].hist(bins=50, figsize=(15, 10))
plt.suptitle("Histograms of All Features")
plt.tight_layout()
plt.savefig('feature_histograms.png')
plt.show()


# ============================================================
# 7. KDE PLOTS
# ============================================================
fig, axes = plt.subplots(3, 2, figsize=(15, 12))
axes = axes.flatten()
colors = ['blue', 'green', 'red', 'cyan', 'navy', 'orange']

for i, (col, color) in enumerate(zip(columns_to_check, colors)):
    sns.kdeplot(bitcoin_df[col], ax=axes[i], fill=True, color=color)
    axes[i].set_title(f'KDE - {col}')

plt.tight_layout()
plt.savefig('feature_kde_plots.png')
plt.show()


# ============================================================
# 8. BOX PLOTS (OUTLIERS)
# ============================================================
fig, axes = plt.subplots(3, 2, figsize=(15, 12))
axes = axes.flatten()

for i, col in enumerate(columns_to_check):
    sns.boxplot(y=bitcoin_df[col], ax=axes[i])
    axes[i].set_title(f'Box Plot - {col}')

plt.tight_layout()
plt.savefig('feature_box_plots.png')
plt.show()


# ============================================================
# 9. CRIAÇÃO DO TARGET (PREVER AMANHÃ)
# ============================================================
X = bitcoin_df.drop(columns=['Close', 'Date'])
y = bitcoin_df['Close'].shift(-1)

X = X.iloc[:-1]
y = y.iloc[:-1]

split_idx = int(len(X) * 0.8)

X_train = X.iloc[:split_idx]
X_test  = X.iloc[split_idx:]
y_train = y.iloc[:split_idx]
y_test  = y.iloc[split_idx:]


# ============================================================
# 10. LOG TRANSFORM + SCALING
# ============================================================
# Log transform
X_train_log = np.log1p(X_train)
X_test_log = np.log1p(X_test)
y_train_log = np.log1p(y_train)
y_test_log = np.log1p(y_test)

# Feature Scaler
scaler_X = StandardScaler()
X_train_scaled = scaler_X.fit_transform(X_train_log)
X_test_scaled  = scaler_X.transform(X_test_log)

# Target Scaler
scaler_y = StandardScaler()
y_train_scaled = scaler_y.fit_transform(y_train_log.values.reshape(-1, 1))
y_test_scaled  = scaler_y.transform(y_test_log.values.reshape(-1, 1))


# ============================================================
# 11. Regressão Linear (BASELINE)
# ============================================================
lr = LinearRegression()
lr.fit(X_train_scaled, y_train_scaled)

y_pred_scaled = lr.predict(X_test_scaled)

# Inversões
y_pred = np.expm1(scaler_y.inverse_transform(y_pred_scaled))
y_true = np.expm1(scaler_y.inverse_transform(y_test_scaled))

mae = mean_absolute_error(y_true, y_pred)
rmse = np.sqrt(mean_squared_error(y_true, y_pred))
r2   = r2_score(y_true, y_pred)

print("\n--- Regressão Linear Baseline ---")
print(f"MAE:  ${mae:,.2f}")
print(f"RMSE: ${rmse:,.2f}")
print(f"R2:   {r2:.2f}\n")


# ============================================================
# 12. PLOT - BASELINE
# ============================================================
plt.figure(figsize=(14,7))
plt.plot(y_true, label='Atual')
plt.plot(y_pred, label='Previsto', linestyle='--')
plt.legend()
plt.title('Regressão Linear: Atual vs Previsto')
plt.grid(True)
plt.savefig('linear_regression_predictions.png')
plt.show()


# ============================================================
# 13. LSTM - PREPARAR SEQUÊNCIAS
# ============================================================
def create_sequences(X, y, steps=60):
    X_seq, y_seq = [], []
    for i in range(len(X) - steps):
        X_seq.append(X[i:i+steps])
        y_seq.append(y[i+steps])
    return np.array(X_seq), np.array(y_seq)

TIME_STEPS = 60
X_train_seq, y_train_seq = create_sequences(X_train_scaled, y_train_scaled)
X_test_seq, y_test_seq   = create_sequences(X_test_scaled, y_test_scaled)


# ============================================================
# 14. LSTM MODEL
# ============================================================
model = Sequential([
    LSTM(50, input_shape=(TIME_STEPS, X_train_scaled.shape[1])),
    Dense(1)
])

model.compile(optimizer='adam', loss='mean_absolute_error')

early_stop = EarlyStopping(monitor="val_loss", patience=5, restore_best_weights=True)

history = model.fit(
    X_train_seq, y_train_seq,
    epochs=100,
    batch_size=32,
    validation_data=(X_test_seq, y_test_seq),
    callbacks=[early_stop],
    verbose=1
)


# ============================================================
# 15. EVALUAR LSTM
# ============================================================
y_pred_scaled_LSTM = model.predict(X_test_seq)
y_pred_LSTM = np.expm1(scaler_y.inverse_transform(y_pred_scaled_LSTM))
y_true_LSTM = np.expm1(scaler_y.inverse_transform(y_test_seq))

mae = mean_absolute_error(y_true_LSTM, y_pred_LSTM)
rmse = np.sqrt(mean_squared_error(y_true_LSTM, y_pred_LSTM))
r2   = r2_score(y_true_LSTM, y_pred_LSTM)

print("\n--- LSTM Evaluation ---")
print(f"MAE:  ${mae:,.2f}")
print(f"RMSE: ${rmse:,.2f}")
print(f"R2:   {r2:.2f}")


# ============================================================
# 16. PLOT LOSS
# ============================================================
plt.plot(history.history['loss'], label='treino')
plt.plot(history.history['val_loss'], label='validação')
plt.legend()
plt.title('LSTM Loss (MAE)')
plt.show()


# ============================================================
# 17. PLOT - LSTM PREVISÃO
# ============================================================
plt.figure(figsize=(14,7))
plt.plot(y_true_LSTM, label='Atual')
plt.plot(y_pred_LSTM, label='Previsto', linestyle='--')
plt.legend()
plt.title('LSTM: Atual vs Previsto')
plt.grid(True)
plt.savefig('lstm_predictions.png')
plt.show()
```

```python
# ============================================
# 1. IMPORTS
# ============================================
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split, cross_val_score
from sklearn.metrics import (
    accuracy_score, precision_score, recall_score, f1_score,
    confusion_matrix, roc_curve, auc
)
from sklearn.ensemble import RandomForestClassifier
from xgboost import XGBClassifier


# ============================================
# 2. CARREGAR O DATASET
# ============================================
df = pd.read_csv("/content/sample_data/coin_Bitcoin.csv", parse_dates=["Date"])
df = df.sort_values("Date").reset_index(drop=True)

df.head()


# ============================================
# 3. CRIAR TARGET PARA CLASSIFICAÇÃO
# ============================================
df_class = df.copy()

df_class["target"] = (df_class["Close"].shift(-1) > df_class["Close"]).astype(int)

df_class = df_class.dropna()


# ============================================
# 4. DEFINIR X E y
# ============================================
X = df_class.drop(columns=["target", "Close", "Date", "Name", "Symbol", "SNo"])
y = df_class["target"]

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, shuffle=False
)

X_train.shape, X_test.shape


# ============================================
# 5. MODELO - RANDOM FOREST
# ============================================
clf_rf = RandomForestClassifier(
    n_estimators=300,
    max_depth=8,
    random_state=42
)
clf_rf.fit(X_train, y_train)

pred_rf = clf_rf.predict(X_test)


# ============================================
# 6. MODELO - XGBOOST
# ============================================
clf_xgb = XGBClassifier(
    n_estimators=300,
    learning_rate=0.05,
    max_depth=5,
    subsample=0.8,
    colsample_bytree=0.8,
    eval_metric="logloss"
)

clf_xgb.fit(X_train, y_train)
pred_xgb = clf_xgb.predict(X_test)


# ============================================
# 7. FUNÇÃO DE AVALIAÇÃO
# ============================================
def avaliar_modelo(nome, y_true, y_pred):
    acc = accuracy_score(y_true, y_pred)
    prec = precision_score(y_true, y_pred)
    rec = recall_score(y_true, y_pred)
    f1 = f1_score(y_true, y_pred)

    print(f"===== {nome} =====")
    print("Acurácia:", acc)
    print("Precisão:", prec)
    print("Recall:", rec)
    print("F1 Score:", f1)
    print("-------------------------")

    return acc, prec, rec, f1


rf_results = avaliar_modelo("Random Forest", y_test, pred_rf)
xgb_results = avaliar_modelo("XGBoost", y_test, pred_xgb)


# ============================================
# 8. MATRIZ DE CONFUSÃO
# ============================================
fig, axes = plt.subplots(1, 2, figsize=(14,5))

sns.heatmap(confusion_matrix(y_test, pred_rf), annot=True, fmt="d", cmap="Blues", ax=axes[0])
axes[0].set_title("Matriz de Confusão - Random Forest")

sns.heatmap(confusion_matrix(y_test, pred_xgb), annot=True, fmt="d", cmap="Greens", ax=axes[1])
axes[1].set_title("Matriz de Confusão - XGBoost")

plt.show()


# ============================================
# 9. CURVA ROC
# ============================================
proba_rf = clf_rf.predict_proba(X_test)[:,1]
proba_xgb = clf_xgb.predict_proba(X_test)[:,1]

fpr_rf, tpr_rf, _ = roc_curve(y_test, proba_rf)
fpr_xgb, tpr_xgb, _ = roc_curve(y_test, proba_xgb)

plt.figure(figsize=(7,5))
plt.plot(fpr_rf, tpr_rf, label=f"RF (AUC = {auc(fpr_rf,tpr_rf):.3f})")
plt.plot(fpr_xgb, tpr_xgb, label=f"XGB (AUC = {auc(fpr_xgb,tpr_xgb):.3f})")
plt.plot([0,1],[0,1],'--', color='gray')
plt.xlabel("FPR")
plt.ylabel("TPR")
plt.title("Curva ROC")
plt.legend()
plt.show()


# ============================================
# 10. IMPORTÂNCIA DAS FEATURES
# ============================================
importances_rf = pd.Series(clf_rf.feature_importances_, index=X.columns)
importances_xgb = pd.Series(clf_xgb.feature_importances_, index=X.columns)

fig, axes = plt.subplots(1, 2, figsize=(16,6))

importances_rf.sort_values().tail(10).plot(kind='barh', ax=axes[0])
axes[0].set_title("Top 5 Features - Random Forest")

importances_xgb.sort_values().tail(10).plot(kind='barh', ax=axes[1])
axes[1].set_title("Top 5 Features - XGBoost")

plt.show()


# ============================================
# 11. VALIDAÇÃO CRUZADA
# ============================================
scores_rf = cross_val_score(clf_rf, X, y, cv=5, scoring="accuracy")
scores_xgb = cross_val_score(clf_xgb, X, y, cv=5, scoring="accuracy")

print("CV Accuracy RF:", scores_rf.mean())
print("CV Accuracy XGB:", scores_xgb.mean())


# ============================================
# 12. TABELA FINAL DE COMPARAÇÃO
# ============================================
comparison = pd.DataFrame({
    "Modelo": ["Random Forest", "XGBoost"],
    "Acurácia": [rf_results[0], xgb_results[0]],
    "Precisão":   [rf_results[1], xgb_results[1]],
    "Recall":     [rf_results[2], xgb_results[2]],
    "F1-score":   [rf_results[3], xgb_results[3]],
    "CV Accuracy": [scores_rf.mean(), scores_xgb.mean()]
})

comparison
```
